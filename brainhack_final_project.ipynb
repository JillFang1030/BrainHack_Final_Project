# Correlation

```python
import pandas as pd
import numpy as np
from scipy.stats import pearsonr
# --- 1. 指定目前的檔案名稱 ---
try:
    # 讀取 Beta 資料 (CSV 格式)
    df_beta_dd = pd.read_csv('Beta_DD.csv')
    df_beta_td = pd.read_csv('Beta_TD.csv')
    print(f"Beta_DD shape: {df_beta_dd.shape}")
    print(f"Beta_TD shape: {df_beta_td.shape}")
    ## see the channel
    print("\nBeta DD Preview:")
    display(df_beta_dd.head())

    # 讀取行為指標資料 (Excel 格式)
    df_behav_dd = pd.read_excel('DD_for_project.xlsx')
    df_behav_td = pd.read_excel('TD_for_project.xlsx')

    print("✅ 所有檔案（CSV 與 Excel）讀取成功")
except Exception as e:
    print(f"❌ 讀取失敗：{e}")
    raise
```

```python
# 2. 欄位標頭清洗 (清除隱形空格，防止抓不到 Character_recog (PR)) ---
for df in [df_beta_dd, df_beta_td, df_behav_dd, df_behav_td]:
    df.columns = df.columns.str.strip()

# 3. 處理 Beta 值 (提取 Channel 6, 16, 21 的 MA 條件)，合併兩組 Beta 資料
beta_all = pd.concat([df_beta_dd, df_beta_td], ignore_index=True)

# 篩選特定頻道與條件 (只選 cond 為 MA)
target_channels = [6, 16, 21]
beta_filtered = beta_all[(beta_all['Channel'].isin(target_channels)) & (beta_all['cond'] == 'MA')]

# 將長格式轉為寬格式：讓 Subject 變回 ID 欄位，Channel 數字變為標題
beta_pivot = beta_filtered.pivot_table(index='Subject', columns='Channel', values='Beta').reset_index()
beta_pivot.columns = ['ID', 'Beta_CH6', 'Beta_CH16', 'Beta_CH21']
```

```python
# --- 4. 合併行為指標 ---
# 合併兩組行為資料
behavior_all = pd.concat([df_behav_dd, df_behav_td], ignore_index=True)

# 提取 ID, C_MC 以及 Character_recog (PR)
# 注意：C_MC 在資料中代表 Morphological Construction
behavior_subset = behavior_all[['ID', 'C_MC', 'Character_recog (PR)']].copy()

# 合併 Beta 資料與行為指標資料 (透過 ID 進行 inner join)
final_df = pd.merge(beta_pivot, behavior_subset, on='ID', how='inner')
# on='ID'： 告訴電腦，當兩張表的 ID 一樣時，就把那一行資料拼在一起。
```

```python
# 20260509
target_channels = [6, 16, 21]

# 篩選特定 Channel 且條件為 'MA' 的資料
beta_filtered = beta_all[beta_all['Channel'].isin(target_channels) & (beta_all['cond'] == 'MA')]

# 將資料從「長格式」轉為「寬格式」，讓每個 Channel 變成獨立的一欄
beta_pivot = beta_filtered.pivot_table(index='Subject', columns='Channel', values='Beta').reset_index()

# 重新命名欄位以便識別
beta_pivot.columns = ['ID', 'Beta_CH6', 'Beta_CH16', 'Beta_CH21']
```

```python
# 挑選行為指標欄位 (要確認欄位名稱與 CSV 一致)
# C_MC 代表 Morphological Construction, Character_recog (PR) 是閱讀識字量的PR值而非原始分屬，以免受到年齡的影響
behavior_subset = behavior_all[['ID', 'C_MC', 'Character_recog (PR)']].copy()

# 確保 ID 型別一致以利合併
beta_pivot['ID'] = beta_pivot['ID'].astype(int)
behavior_subset['ID'] = behavior_subset['ID'].astype(int)

# 合併表格
final_df = pd.merge(beta_pivot, behavior_subset, on='ID', how='inner')

# 處理缺失值 (如有 NA 則刪除該橫列)
final_df = final_df.dropna(subset=['C_MC', 'Character_recog (PR)'])
```

```python
def run_correlation(df, channels, behaviors):
    results = []
    for ch in channels:
        for bh in behaviors:
            # 去除該對組的缺失值
            valid_data = df[[ch, bh]].dropna()
            r_val, p_val = pearsonr(valid_data[ch], valid_data[bh])
            results.append({
                'Channel': ch,
                'Behavior': bh,
                'Pearson_r': round(r_val, 3),
                'p-value': round(p_val, 4),
                'Significant': '*' if p_val < 0.05 else ''
            })
    return pd.DataFrame(results)

# 執行分析
analysis_results = run_correlation(
    final_df,
    ['Beta_CH6', 'Beta_CH16', 'Beta_CH21'],
    ['C_MC', 'Character_recog (PR)']
)

print("--- 相關分析結果 ---")
print(analysis_results)
```

AI advice: 全體不顯著（$p > 0.05$）的結果是非常常見的。這並不代表你的專案失敗了，反而給了你更深層的分析動機。

以下是為何會這樣的可能，以及如何「拯救」這個結果：
1. 為什麼結果是「不顯著」？（診斷原因）混合了不同群體

*   你現在是把 DD (閱讀障礙組) 和 TD (典型發展組) 合在一起算。在神經科學中，這兩組人的腦活動模式往往是相反或不同的。例子：可能 TD 組是強烈正相關，但 DD 組是強烈負相關，兩組一加起來，剛好就把相關性「抵消」掉了，導致 $r$ 接近 0。
*   樣本量 (Sample Size)：如果受試者人數較少，統計效力（Power）不足，很難達到 $p < 0.05$。功能定位：也許 Channel 6, 16, 21 不是負責「構詞（C_MC）」的最核心區域，這也是一個重要的發現（Null finding）。


2. 如何讓專案變「可行」且符合課程要求？
*   這門課教你 Machine Learning 和 Data Analysis，重點不在於結果一定要顯著，而在於你如何挖掘數據。
*   建議動作：分組分析 (Sub-group Analysis)這才是這門課真正的精華。我們應該分別看 DD 組和 TD 組的相關性。

```python
# 分別看 DD 組和 TD 組的相關性
# 1. 找出 DD 組和 TD 組各自的 ID 列表
dd_ids = df_behav_dd['ID'].unique()
td_ids = df_behav_td['ID'].unique()

# 2. 從最終合併的大表 (final_df) 中把兩組人拆開
df_dd_only = final_df[final_df['ID'].isin(dd_ids)]
df_td_only = final_df[final_df['ID'].isin(td_ids)]

# 3. 分別執行分析
print("========== DD (閱讀障礙組) 相關分析 ==========")
results_dd = run_correlation(df_dd_only,
                             ['Beta_CH6', 'Beta_CH16', 'Beta_CH21'],
                             ['C_MC', 'Character_recog (PR)'])
print(results_dd)

print("\n========== TD (典型發展組) 相關分析 ==========")
results_td = run_correlation(df_td_only,
                             ['Beta_CH6', 'Beta_CH16', 'Beta_CH21'],
                             ['C_MC', 'Character_recog (PR)'])
print(results_td)
```
